# Agentic Scientific Analysis Starter

This notebook demonstrates a minimal workflow for scientific analysis with a lightweight agentic loop.


## 1) Create synthetic experiment data

In a real project, replace this with your instrument, simulation, or public dataset loading code.


```python
from random import gauss, seed
from statistics import mean, stdev

seed(7)
control = [gauss(10.0, 1.0) for _ in range(30)]
treatment = [gauss(11.2, 1.1) for _ in range(30)]

data = {"control": control, "treatment": treatment}
data
```

## 2) Define analysis tools

Treat each tool as a callable function the agent can use.


```python
def summarize_groups(dataset):
    return {
        group: {
            "n": len(values),
            "mean": round(mean(values), 3),
            "stdev": round(stdev(values), 3),
        }
        for group, values in dataset.items()
    }

def mean_difference(dataset):
    return round(mean(dataset["treatment"]) - mean(dataset["control"]), 3)
```

## 3) Add a minimal agent loop

This mock planner simulates how an LLM/agent would choose tools.
Swap `mock_planner` with a real model client when API access is available.


```python
def mock_planner(question):
    q = question.lower()
    if "effect" in q or "difference" in q:
        return ["summarize_groups", "mean_difference"]
    return ["summarize_groups"]

TOOLS = {
    "summarize_groups": summarize_groups,
    "mean_difference": mean_difference,
}

def run_agent(question, dataset):
    plan = mock_planner(question)
    results = {}
    for step in plan:
        results[step] = TOOLS[step](dataset)
    return {"question": question, "plan": plan, "results": results}
```

```python
run_agent("What is the treatment effect compared to control?", data)
```
